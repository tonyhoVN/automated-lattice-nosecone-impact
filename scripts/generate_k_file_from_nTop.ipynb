Generate Raw k file from nTOP
===

```python
from ntopcl import *
import os
```

```python
main_folder = os.path.abspath(os.path.join(os.path.dirname('__file__'), ".."))
nTop_folder = os.path.join(main_folder, "nTop_files")
cad_folder = os.path.join(main_folder, "CAD_files")

nTopFilepath = os.path.join(nTop_folder, "Lattice_nose_cone.ntop")
model_step_file = os.path.join(cad_folder, "NoseCone.STEP")
inputTemplate = os.path.join(nTop_folder, "input_template.json")
inputFilePath = os.path.join(nTop_folder, "input_template.json")      #JSON input file name to be saved as
outputFilePath = os.path.join(nTop_folder, "output_template.json")       #JSON output file name to be saved as

dyna_folder = os.path.join(main_folder, "DYNA_files")
input_k_file = os.path.join(dyna_folder, "lattice_nose_cone_pre.k")
output_modify_k_file = os.path.join(dyna_folder, "lattice_nose_cone_pre.k")
output_stl_file = os.path.join(cad_folder, "lattice_nose_cone.stl")
```

### Create JSON input, output template

```python
json_template(nTopFilepath)
```

### Change input values

```python
# Create json input file with new inputs
Inputs = {
    "Input Path": model_step_file,
    "Output Path k file": input_k_file,
    "Output path STL": output_stl_file,
    "Thickness": 1.0,
    "UVW Count": [4, 6, 1],
}

edit_json_input(inputTemplate, inputFilePath, Inputs)
```

### Run nTop via client server

```python
# Run ntopcl to get json output file
run_json(False, True, inputFilePath, outputFilePath, nTopFilepath)
output_result = get_json_output(outputFilePath)
print("Number of cells: %d" %(output_result))
```
